```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/Users/abhijain/Documents/AI-ML-class-practice-/data/Titanic-Dataset.csv")
mpg = pd.read_csv("/Users/abhijain/Documents/AI-ML-class-practice-/data/mpg.csv")
```

```python
print(df.shape,mpg.shape)
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x="Fare")
plt.title("fare value ")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x="Fare",y="Age")
plt.title("fare and Age ")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x="Fare",y="Age",hue="Sex")
plt.title("fare , age and sex ")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x="Fare",bins=25)
plt.title("fare value ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x="Age",bins=25)
plt.title("age ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.countplot(data=df,x="Survived")
plt.title("fare value ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.countplot(data=df,x="Survived",hue="Sex")
plt.title("servived and sex ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.histplot(data=mpg,x="mpg",bins=25)
plt.title("mpg data ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.countplot(data=mpg,x="origin")
plt.title("cars origin ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(data=df,y="Fare")
plt.title(" fare ")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(data=df,x="Survived",y="Fare")
plt.title(" fare and survived ")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(data=df,x="Survived",y="Age")
plt.title(" age ")
plt.grid()
plt.show()
```
