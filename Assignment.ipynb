```python
s1 = [72,85,60,91]
s2 = [45,30,98,55]
s3 = [88,66,74,100]
f = np.array((s1,s2,s3))
old = np.array([['62','71','55','80'],
                ['49','58','66','73']])
print("no of student",f.shape[0])
print("no of subjects",f.shape[1])
print("total no of marks entries ",np.size(f))
print("total sum ",np.sum(f))
print(f.astype('float32'))
old = old.astype("int32")
print("converted into int ",old)
f = np.vstack((f,old))
print("combined both array",f)
```
